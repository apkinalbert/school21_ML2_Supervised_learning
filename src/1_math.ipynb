## 1. Analytical Solution to the Regression Problem (Vector Form)

**Setup.** Let:

- $X \in \mathbb{R}^{n \times d}$ be the feature matrix, where $n$ is the number of samples and $d$ is the number of features.
- $y \in \mathbb{R}^{n}$ be the target vector.
- $w \in \mathbb{R}^{d}$ be the vector of model parameters.

The linear regression model is

$$
\hat{y} = Xw
$$

**Loss function.** The goal is to minimize the Mean Squared Error. Since the factor $\frac{1}{n}$ does not affect the minimizer, it is enough to minimize the equivalent objective

$$
L(w) = \|y - Xw\|_2^2
$$

This objective measures the squared Euclidean norm of the residual vector $y - Xw$.

**Step 1: Expand the squared norm.**

Using the definition of the Euclidean norm,

$$
L(w) = (y - Xw)^T (y - Xw)
$$

Now expand the transpose of the difference:

$$
L(w) = (y^T - w^T X^T)(y - Xw)
$$

Multiply the two factors term by term:

$$
L(w) = y^T y - y^T Xw - w^T X^T y + w^T X^T X w
$$

Notice that $y^T Xw$ is a scalar. Therefore, it is equal to its transpose:

$$
y^T Xw = (y^T Xw)^T = w^T X^T y
$$

So the two middle terms are equal, and the expression simplifies to

$$
L(w) = y^T y - 2y^T Xw + w^T X^T X w
$$

**Step 2: Differentiate with respect to $w$.**

Use the standard matrix derivative rules:

- $\nabla_w (w^T A w) = 2Aw$ when $A$ is symmetric.
- $\nabla_w (b^T w) = b$.
- The derivative of a constant is zero.

Here, $A = X^T X$, and this matrix is symmetric because

$$
(X^T X)^T = X^T X
$$

Therefore,

$$
\nabla_w L(w) = 2X^T X w - 2X^T y
$$

**Step 3: Set the gradient equal to zero.**

At the minimum, the gradient must be zero:

$$
\nabla_w L(w) = 0
$$

Hence,

$$
2X^T X w - 2X^T y = 0
$$

Rearranging gives

$$
X^T X w = X^T y
$$

This equation is called the **normal equation**.

**Step 4: Solve for the optimal weight vector.**

Multiply both sides on the left by $(X^T X)^{-1}$:

$$
(X^T X)^{-1} X^T X w = (X^T X)^{-1} X^T y
$$

Since $(X^T X)^{-1} X^T X = I$, we obtain

$$
w = (X^T X)^{-1} X^T y
$$

This is the analytical, or closed-form, solution to the linear regression problem in vector form.

## 2. Effect of L1 and L2 regularization on the analytical regression solution
Base regression problem (no regularization)

We start from standard least squares:

$$
L(w)=∥Xw−y∥_2^2
$$	​


Vector solution (normal equation):

$$
w=(X^TX)^{−1}X^Ty
$$

This solution assumes that: $X^T X$ is invertible

### 2.1 L2 regularization (Ridge regression)

Modified loss function

We add penalty on the magnitude of weights:

$$
L(w)=∥Xw−y∥_2^2 +\lambda∥w∥_2^2
$$​	​


where:

- $\lambda \ge 0$ — regularization strength

- $||w||_2^2 = w^T w$

Expand gradient

We differentiate:

- from least squares:
$$
∇_w∥Xw−y∥_2^2=2X^TXw−2X^Ty
$$

- from regularization:

$$
∇_w(λw^Tw)=2λw
$$

So total gradient:

$$
∇_wL(w)=2X^TXw−2X^Ty+2λw
$$

Set gradient to zero
$$
2X^TXw−2X^Ty+2λw=0
$$

Divide by 2:
$$
X^TXw+λw=X^Ty
$$

Factor w:
$$
(X^TX+λI)w=X^Ty
$$

Closed-form solution (Ridge)
$$
w=(X^TX+λI)^{−1}X^Ty
$$

What changed compared to ordinary regression?
- Matrix becomes always invertible (if $\lambda > 0$)
- Solution becomes more stable numerically
- We shrink weights toward zero but do not force sparsity

$$
X^T X + \lambda I =
\begin{pmatrix}
a_{11}+\lambda & a_{12} & \dots \\
a_{21} & a_{22}+\lambda & \dots \\
\vdots & \vdots & \ddots
\end{pmatrix}
$$

### 2.2 L1 regularization (Lasso)

Modified loss function

$$
L(w)=∥Xw−y∥_2^2 +\lambda∥w∥_1
$$​	​

where:
$
∥w∥_1=\sum_{i=1}^{d} ∣w_i|​
$

Key difficulty

Unlike L2:

$|w_i|$ is not differentiable at $w_i = 0$
so we cannot derive a clean closed-form solution

We use subgradient:

$$
\partial |w_i| =
\begin{cases}
+1, & w_i > 0 \\
-1, & w_i < 0 \\
[-1,1], & w_i = 0
\end{cases}
$$
	​

Optimality condition
$$
2X^T(Xw−y)+λ⋅∂∥w∥_1=0
$$

This equation cannot be solved analytically in closed form in general.

Resulting solution form

Instead of a matrix inverse, solution is:

- obtained via iterative optimization
- has sparsity property

Meaning: $w_i=0$ for many features

### 2.3 Why are there many weights equal to 0 after the model is fit?

Many weights become equal to zero in L1-regularized regression (Lasso) because the derivative of the regularization term $|w|$ is not unique at $w=0$. Its subgradient at zero is:

The interval $[-1,1]$ means that zero can satisfy the optimality condition for a range of gradient values. As a result, the optimization process tends to set many coefficients exactly to zero, producing a sparse model and performing implicit feature selection.

## Linear regression, Ridge, etc. possible to fit nonlinear dependencies.

Linear models such as Linear Regression, Ridge, and Lasso can model nonlinear dependencies by transforming the original features into nonlinear features before training.

Instead of using the original input vector $x$, we apply a feature transformation:
$$
\phi(x)
$$

and the model becomes:
$$
\hat y = w^T \phi(x)
$$

For example, for a one-dimensional input $x$, we can create polynomial features:
$$
\phi(x) = [1, x, x^2, x^3]
$$

Then the regression model is:
$$
\hat y = w_0 + w_1 x + w_2 x^2 + w_3 x^3
$$

Although the model is nonlinear with respect to the input variable $x$, it is still linear with respect to the parameters $w$. Therefore, the same analytical and optimization methods used in Linear Regression, Ridge, and Lasso can still be applied.

Ridge and Lasso regularizations can also be used with nonlinear feature transformations to prevent overfitting when the number of generated features becomes large.
